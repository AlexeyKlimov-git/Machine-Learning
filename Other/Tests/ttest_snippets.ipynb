```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
group1 = np.random.normal(loc=50, scale=10, size=300) # mu = 50, sd = 10
group2 = np.random.normal(loc=55, scale=10, size=300)
```

```python
t_statistic, p_value = stats.ttest_ind(group1, group2)
t_statistic, p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("Reject null hypothesis: There is a significant difference btw the two groups")
else:
    print("Fail to reject null hypothesis: No significant difference between the two groups")
```

```python
plt.figure()
plt.hist(group1, alpha=0.5, label="Group 1", bins=100)
plt.hist(group2, alpha=0.5, label="Group 2", bins=100)
plt.title("Distribution of two groups")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# Calculate sample size
import statsmodels.stats.api as sms

effect_size = 0.5
alpha = 0.05
power = 0.80
```

```python
sample_size_per_group = sms.TTestIndPower().solve_power(effect_size=effect_size, alpha=alpha, power=power)
```

```python
print(f"Optimal sample size per group: {int(sample_size_per_group)}")
```
